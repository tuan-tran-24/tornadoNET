# Extract Frames from Video

## Function

```python
import cv2
import sys
import argparse
from pathlib import Path

def extract_frames(path_in, path_out, step):
    path_in = Path(path_in)
    path_out = Path(path_out)
    path_out.mkdir(parents=True, exist_ok=True)

    video_name = path_in.stem
    vidcap = cv2.VideoCapture(str(path_in))
    
    count = 0
    while True:
        vidcap.set(cv2.CAP_PROP_POS_MSEC, count * 1000)    
        success, image = vidcap.read()

        if not success:
            break
            
        frame_out = path_out / f"{video_name}_frame_{count + 1}.jpg"    
        cv2.imwrite(str(frame_out), image)    
        
        count = count + step
        
    vidcap.release()
```

## Implementation

```python
# inputs
step = 1 #seconds
path_in = r'/.../Videos/2011/Joplin1R 6 14'
path_out = r'/.../Frames/2011/Joplin1R 6 14'

extract_frames(path_in, path_out, step)
```
